````python
from typing import Any

from dotenv import load_dotenv
load_dotenv()
````

````python
from anthropic import Anthropic
client = Anthropic()
````

````python
model = "claude-sonnet-4-0"
````

````python
your_json = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "Дай JSON, содержащий в поле description ``` в середине текста"
        },
        {
            "role": "assistant",
            "content": "```json"
        }
    ],
    stop_sequences=["```"]
)
````
